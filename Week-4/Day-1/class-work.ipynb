Performing different operation on Tensor just like numpy Array.

```python
t1 = torch.tensor([2., 4., 6.])
t2 = torch.tensor([3., 5., 8.])

sub = t2-t1
add = t2+t1
mul = t2*t1
div = t2/t1

# subtract, add, multiple, division
print(sub)
print(add)
print(mul)
print(div)
```

```python
# aggregate methods
total = sub + mul + add + div
print(total.mean())
print(total.max())
print(total.min())
```

```python
# item method
print(total.max().item())
print(total.argmax().item())
```

**Multiply inputs by weights**
weight: tells how strongly one input affects a neuron.

```python
student = torch.tensor([9.0, 89.0])
weight = torch.tensor([0.3, 0.05])

result = student * weight
print(result)
print(result.sum())
```

```python
bias = 2.0

res = result.sum() + bias
activation = torch.relu(res)

print("Weighted sum plus bias:", res.item())
print("ReLU activation:", activation.item())
```

```python
# logit

logit = torch.tensor(1.5)
probability = torch.sigmoid(logit)          # sigmoid convert any values in range 0 and 1.
prediction = (probability >= 0.5).int()     # 0.5 threshold point
print(probability,":", prediction.item())
```

```python
# Single Linear Neural Network
one_row = torch.tensor([25.2, 90.2])
one_layer = nn.Linear(2, 1)
one_score = one_layer(one_row)
print(one_row.shape,":",one_score)
```

```python
hidden_layer = nn.Linear(2, 4)
hidden = hidden_layer(one_row)
hidden = torch.relu(hidden)
output_layer = nn.Linear(4, 1)

print("Hidden layer:",hidden.shape,"||", output_layer(hidden).shape,"=>",output_layer(hidden))
```

**Binary Cross-entropy**: measures how strongly prediction agree with its 0 or 1 (target).

**Loss gradients**: --!!

```python

weight = torch.tensor(6., requires_grad=True)
entropy_loss = (weight-4).square()
entropy_loss.backward()

with torch.no_grad():
    weight -= 0.25 * weight.grad

print("weight:", weight.item(),"|| Entropy Loss:", entropy_loss.item())
```

Split the data before learning

```python
from sklearn.model_selection import train_test_split  # Split rows and labels together.

X_train_df, X_holdout, y_train_df, y_holdout = train_test_split(  # Reserve 30% for validation and testing.
    X, y, test_size=0.30, random_state=42, stratify=y  # Repeat the split and keep class proportions close.
)
X_valid_df, X_test_df, y_valid_df, y_test_df = train_test_split(  # Divide the held-out rows in half.
    X_holdout, y_holdout, test_size=0.50, random_state=42, stratify=y_holdout  # Keep labels balanced across both halves.
)
print(f"X_train: {len(X_train_df)},\nX_valid{len(X_valid_df)},\nX_test{len(X_test_df)}")  # Show train, validation, and test counts.
```

```python
def to_float_tensor(frame):  # Convert numeric labels for PyTorch.
    values = frame.to_numpy(dtype=np.float32, copy=True)  # Keep the one-column shape.
    return torch.from_numpy(values)  # Wrap the NumPy array as a tensor.

y_train = to_float_tensor(y_train_df)  # Convert training labels.
y_valid = to_float_tensor(y_valid_df)  # Convert validation labels.
y_test = to_float_tensor(y_test_df)  # Convert test labels.
print(y_train.shape)  # Check that each student has one label.
```

Standarize the input features

```python
from sklearn.preprocessing import StandardScaler  # Reuse the Module 3 scaler.

scaler = StandardScaler().fit(X_train_df)  # Learn scaling from training rows only.
X_train = torch.from_numpy(scaler.transform(X_train_df))  # Scale training features.
X_valid = torch.from_numpy(scaler.transform(X_valid_df))  # Reuse the same scaling.
X_test = torch.from_numpy(scaler.transform(X_test_df))  # Reuse the same scaling.
X_train, X_valid, X_test = X_train.float(), X_valid.float(), X_test.float()  # Match model dtype.
mean = torch.tensor(scaler.mean_, dtype=torch.float32)  # Keep the means for inference.
scale = torch.tensor(scaler.scale_, dtype=torch.float32)  # Keep the scales for inference.
print(X_train.mean(dim=0))  # Check that training feature means are near zero.
print(X_train.std(dim=0))  # Check that training feature scales are near one.
```

```python
from sklearn.dummy import DummyClassifier  # Build a majority-class baseline.
from sklearn.metrics import accuracy_score  # Count the fraction of correct labels.

baseline = DummyClassifier(strategy='most_frequent')  # Always predict the common class.
baseline.fit(X_train_df, y_train_df)  # Learn the common label from training data.
baseline_predictions = baseline.predict(X_valid_df)  # Predict on validation rows.
baseline_valid = accuracy_score(y_valid_df, baseline_predictions)  # Score the baseline.
print(baseline_predictions[0], round(baseline_valid, 3))  # Show its class and accuracy.
```
