# File Handling

File handling is essential for many programs. Below are concise notes and practical examples.

- **Key function:** open(filename, mode)
- **Modes (text):** r (read), a (append), w (write), x (create)
- **Binary/text flag:** add `b` for binary, `t` for text (default)
- **Use `with`** to auto-close files; otherwise call `close()`
- **Common ops:** read(), read(n), readline(), iterate lines, write(), remove files/folders

Follow the code cell below for short, runnable examples with guiding comments.

```python
# --- Examples: reading ---
# simple read (text mode is default)
with open('TestFile.txt', 'r') as f:
    print(f.read())
```

```python
# read only first 5 characters
with open('TestFile.txt', 'r') as f:
    print(f.read(6))
```

```python
# readline and multiple calls
with open('TestFile.txt', 'r') as f:
    print(f.readline())
    print(f.readline())
```

```python
# iterate file line-by-line (memory friendly)
with open('TestFile.txt', 'r') as f:
    for line in f:
        print(line.strip())
```

```python
# --- Examples: writing ---
# append to a file (creates if not exists)
with open('TestFile.txt', 'a') as f:
    f.write('\nNow the file has more content!\nThe brown fox jumps over the lazy dog is the general sentence which is use for grip whole keybord keys.')
```

```python

# overwrite entire file (danger: deletes previous content)
with open('TestFile_overwrite.txt', 'w') as f:
    f.write('Woops! I have deleted the content!')
```

```python

# create a file; using 'x' raises if file exists
try:
    with open('TestFile.txt', 'x') as f:
        f.write('This is ')
except FileExistsError:
    print('TestFile.txt already exists')
```

```python
# inheritance
class Animal:
    def eat(self):
        return "Eating..."

class Dog(Animal):
    pass    # use parent behaviour

print(Dog().eat())
```

```python
# inheritance with constructor
class Animal:
    def __init__(self, name):
        self.name = name
    def eat(self):
        return "Eating..."

class Dog(Animal):
    pass    # use parent behaviour

print(Dog("spike").eat())
```

```python
# polymorphism: same methods but different object
class Animal:
    # parent class
    def speak(self):
        return "Sound"
    
# child class
class Dog(Animal):
    def speak(self):
        return "Bark"

# child class
class Cat(Animal):
    def speak(self):
        return "Meow"

# calling methods to their class: here methods `speak` same but different classes
print(Dog().speak())
print(Cat().speak())
```

```python
# SQLite

import sqlite3

# Connect to the database (creates tasks.db if it doesn't exist)
db = sqlite3.connect("tasks.db")

# Create the tasks table if it doesn't already exist
db.execute("""
    CREATE TABLE IF NOT EXISTS tasks (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        title TEXT NOT NULL
    )
""")
rows = db.execute(
    """SELECT title FROM tasks;"""
)
for r in rows:
    print(r[0])
db.commit()
db.close()
```
