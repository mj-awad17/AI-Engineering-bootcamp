Practice: Create a grade array

Make a numpy array with four grades: 65, 75, 85, 95. Print its shape and average.
Use explanatory python comments and print/check the output.

```python
grades = np.array([65, 75, 85, 95])

# average calculating using custome logic
avge_cust_logic = 0
for i in grades:
    avge_cust_logic += i

# using numpy method
avge_method = grades.mean()
# using numpy methods without using mean method
avge_method_no_mean = grades.sum()/len(grades)

print(f"Shape of Grades: {grades.shape}")

print(f"Average grades(using logic): {avge_cust_logic}")
print(f"Average grades(numpy method[mean]): {avge_method}")
print(f"Average grades(numpy method without using mean): {avge_method_no_mean}")
```

Practice: Create a 2x3 array, print its flattened values and transposed shape.

```python
arr = np.arange(1, 7)
print(f"Original array:\n\t{arr}")
n_arr = arr.reshape(2,3)
print(f"Reshape array:\n{n_arr}")
f_arr = n_arr.flatten()
print(f"Flatten array:\n\t{f_arr}")
print(f"Original array shape: {n_arr.shape}")
# transpose the array
n_arrT = n_arr.T
print(f"After T array shape: {n_arrT.shape}")
```

```python
# mathematical operations on numpy array
a = np.arange(1, 4)
b = np.arange(4, 9, 2)
# print(a, b)
print(np.add(a,b))
print(np.subtract(a,b))
print(np.divide(a,b))
print(np.multiply(a,b))
```

```python
# let perform on 2D array
a = np.arange(1, 16, 2).reshape(2,4)
b = np.arange(16, 32, 2).reshape(2,4)
print(a)
print("-"*25)
print(b)
print("-"*25)
print(np.add(a,b),"\n")
print(np.subtract(a,b),"\n")
print(np.divide(a,b),"\n")
print(np.multiply(a,b),"\n")
print("-" *20)
print(np.square(a))
print(np.sqrt(b))
```

```python
# dot product
feature = np.arange(1, 16, 2)
weights = np.arange(16, 32, 2)

score = np.dot(feature, weights)
print("Weighted Score:", score)
print("Total length of elements:\n","Feature:",len(feature),"\n Weights:",len(weights))
f = feature.reshape(4,2)
w = weights.reshape(2,4)
s = np.dot(f, w)
print("Weighted Score(2D):\n",s)
```

Practice: Elementwise vs dot
For arrays `[10, 20, 30]` and `[40, 50, 60]`, show elementwise product and dot product.

```python
arr1 = np.array([10, 20, 30])
arr2 = np.array([40, 50, 60])

# elementwise product
arr = arr1 @ arr2
arr_e = np.multiply(arr1, arr2)
print("Elementwise product:",arr_e)

# dot product
arr_d = np.dot(arr1, arr2)
print("Dot product:",arr_d)
```

```python
import numpy as np
a = np.arange(1, 16, 2).reshape(2,4,1)
print(a)
print(a.sum())
print(np.mean(a))
print(np.median(a))
# print(np.mode(a, axis=1))
print(round(np.std(a), 2))
print(round(np.min(a)))
print(round(np.max(a)))
# z-score = mean - std
# z-score= lower is better hower formula is above one and it also tell use the spread of data
mean = np.mean(a)
std = np.std(a)
print(list(np.round((a - mean)/std)))
```
