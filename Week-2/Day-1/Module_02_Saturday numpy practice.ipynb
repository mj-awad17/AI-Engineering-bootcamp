## Topic 01: Python Lists vs NumPy Arrays: Correctness and Runtime

**What is it?** A benchmark measures how long equivalent computations take on a given machine and input size.

**Why use it?** For many numerical operations, NumPy performs work in optimized compiled loops, reducing Python per-element overhead.

**Important functions and methods**
- `Python list` — Flexible container; may hold mixed types.
- `np.ndarray` — Fixed-dtype numerical array designed for vectorization.
- `time.time()` — Read the clock and calculate elapsed seconds.
- `round()` — Show a shorter decimal runtime.
- `np.allclose()` — Check numerical results, allowing small float differences.

**Remember:** Timing varies by machine, package version, array size and first-run overhead; never promise a universal speedup.

### Code Example 1 — Same mathematical task and speed comparison
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Work with numerical arrays.
import time  # Measure how long the code takes.
numbers = list(range(100000))  # Make 100,000 numbers.
arr = np.array(numbers)  # Convert list into an array.
start = time.time()  # Start timing the Python list.
list_result = [x + 5 for x in numbers]  # Add 5 to each value.
list_time = time.time() - start  # Calculate list time.
start = time.time()  # Start timing the NumPy array.
array_result = arr + 5  # Add 5 to all values at once.
numpy_time = time.time() - start  # Calculate NumPy time.
print("List:", round(list_time, 5))  # Show list time.
print("NumPy:", round(numpy_time, 5))  # Show array time.
```

## Topic 08: Array comparisons, boolean masks and where

**What is it?** Comparisons return True/False arrays that can filter or conditionally replace numerical values.

**Why use it?** Masks help select students above a threshold and detect invalid values in model features.

**Important functions and methods**
- `a > threshold` — Create a boolean mask
- `a[mask]` — Keep matching entries
- `np.where(condition,a,b)` — Choose values conditionally
- `np.array_equal(a,b)` — Check entire-array equality
- `np.any(mask)/np.all(mask)` — Test whether any/all are True

**Remember:** Use `&` and `|` for elementwise conditions, with parentheses around each comparison. Python `and` is not arraywise.

### Code Example 1 — Boolean comparison of scores
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
marks = np.array([42, 85, 61, 90])  # Student marks.
mask = marks >= 60  # Compare every mark.
print(mask)  # Booleans for pass threshold.
print(marks[mask])  # Keep only passing marks.
```

### Code Example 2 — Use where to label a threshold
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
ages = np.array([12, 18, 21])  # Three ages.
labels = np.where(ages >= 18, "Adult", "Minor")  # Choose labels.
print(labels)  # Return text array.
```

### Code Example 3 — Compare arrays for equality
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
a = np.array([2, 4, 6])  # First array.
b = np.array([2, 5, 6])  # Second array.
print(a == b)  # Compare each corresponding value.
print(np.array_equal(a, b))  # Whole-array equality.
```

### Code Example 4 — Any and all validation
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
scores = np.array([45, 75, 110])  # Data may contain invalid marks.
invalid = (scores < 0) | (scores > 100)  # Find invalid entries.
print("Any invalid:", np.any(invalid))  # At least one bad value.
print("All valid:", np.all(~invalid))  # Every entry passes validation.
```

## Topic 09: Broadcasting: operations on compatible shapes

**What is it?** Broadcasting lets NumPy perform operations between arrays of differing but compatible shapes.

**Why use it?** Use broadcasting to add a bias or normalize each feature column without manually looping over every record.

**Important functions and methods**
- `a + scalar` — Apply one scalar to every entry
- `matrix + row` — Add a row vector to each matrix row
- `arr.shape` — Check compatible dimensions
- `np.newaxis` — Insert a length-one axis
- `np.broadcast_shapes(a,b)` — Check if shapes can broadcast

**Remember:** Broadcasting is not arbitrary: trailing dimensions must be equal or one of them must be 1.

### Code Example 1 — Add a scalar to every score
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
marks = np.array([50, 70, 85])  # Three scores.
print(marks + 5)  # Broadcast scalar 5 to all elements.
```

### Code Example 2 — Add per-subject bonuses
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
marks = np.array([[60, 70], [80, 90]])  # Two students, two subjects.
bonuses = np.array([5, 10])  # Bonus for each column.
print(marks + bonuses)  # Broadcast bonuses to both rows.
```

### Code Example 3 — Column vector times row vector
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
a = np.array([[1], [2], [3]])  # Shape (3,1).
b = np.array([10, 20])  # Shape (2,).
print(a * b)  # Result broadcasts to shape (3,2).
```

### Code Example 4 — Center each feature separately
Run the commented code cell. Compare the actual output with your prediction.

## Topic 10: Missing values, infinity, saving arrays and AI readiness

**What is it?** Numerical data may contain NaN (missing/unavailable) and infinity; validate them before calculations.

**Why use it?** Save intermediate arrays for reproducible experiments and ensure model input values are finite.

**Important functions and methods**
- `np.isnan()` — Locate missing numeric NaN values.
- `np.isfinite()` — Check for neither NaN nor infinity.
- `np.nan_to_num()` — Replace NaN/infinity with numeric values.
- `np.save()/np.load()` — Persist and restore .npy arrays.
- `np.random.default_rng()` — Make a reproducible random generator.

**Remember:** Runtime files in Colab are temporary; use Google Drive or download the files if persistence beyond a reset is required.

### Code Example 1 — Detect missing values
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
x = np.array([1.0, np.nan, 3.0])  # Data with missing number.
print(np.isnan(x))  # Boolean mask of missing entries.
print(np.isfinite(x))  # Check valid finite values.
```

### Code Example 2 — Safely fill non-finite numbers
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
x = np.array([1., np.nan, np.inf])  # Three numerical cases.
clean = np.nan_to_num(x, nan=0., posinf=999.)  # Choose replacements.
print(clean)  # Display finite result.
```

### Code Example 3 — Save and reload one array
Run the commented code cell. Compare the actual output with your prediction.

```python
import numpy as np  # Load NumPy.
marks = np.array([80, 90, 95])  # Data to store.
np.save("marks.npy", marks)  # Write a NumPy binary file.
loaded = np.load("marks.npy")  # Reopen the saved array.
print(loaded.tolist())  # Verify saved content.
```

### Code Example 4 — Check input for model training
