```python
# make new marks array using numpy
# alias `np` already run we use this
marks = np.array([39, 29, 49, 59])
print(marks)

# check the dimension
print(f"Dimension of marks: {marks.ndim}")

# check the shape of numpy array
print(f"Shape: {marks.shape}")

# check the data type
print(f"Datatype: {marks.dtype}")

# change the data type using astype
marks_int32 = marks.astype('int32')
print(f"After change the datatype: {marks_int32.dtype}")
```

```python
# create regular even numbers
values = np.arange(2, 11, 2)
print(f"Even numbers: {values}")
print(f"Shape: {values.shape}\nDataType: {values.dtype}")
```

```python
# zero filled array
attendence = np.zeros(10, dtype='int32')
# replace item with other
attendence[0] = 1
attendence[4] = 1
attendence[6] = 1
print(f"Attendence for one student: {attendence}")
```

```python
# datatype converstion
print(f"Marks: {marks}")

float_marks = marks.astype('float64')
print(f"Float Marks: {float_marks}")
```

Practice: Create a grade array

Make a numpy array with four grades: 65, 75, 85, 95. Print its shape and average.
Use explanatory python comments and print/check the output.

```python
grades = np.array([65, 75, 85, 95])

# average calculating using custome logic
avge_cust_logic = 0
for i in grades:
    avge_cust_logic += i

# using numpy method
avge_method = grades.mean()
# using numpy methods without using mean method
avge_method_no_mean = grades.sum()/len(grades)

print(f"Shape of Grades: {grades.shape}")

print(f"Average grades(using logic): {avge_cust_logic}")
print(f"Average grades(numpy method[mean]): {avge_method}")
print(f"Average grades(numpy method without using mean): {avge_method_no_mean}")
```
